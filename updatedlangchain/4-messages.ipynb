### Messages

these are the fundamental unit of context for models in langchain 


it contains:
Role,
Content,
Metadata.

```python
import os
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")

model=init_chat_model("groq:openai/gpt-oss-120b")
```

```python
model.invoke("Please tell me what is AI")
```

### Text Prompts

```python
model.invoke("Please tell me what is AI")
```

### Message Prompts

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

messages=[
    SystemMessage("You are the Poetry expert"),
    HumanMessage("Write a poem on AI")
]
response=model.invoke(messages)
response.content
```

```python
system_msg=SystemMessage("You are helpful coding assistant.")
messages=[
    system_msg,
    HumanMessage("How do i create a Rest API?")
]
response=model.invoke(messages)
print(response.content)
```

```python
#detailed info to LLM through System Message

from langchain.messages import SystemMessage, HumanMessage

system_msg=SystemMessage("""
You are senior python developer with expertise in web frameworks.
Always provide code examples and explain you reasoning.
Be concise but through in your explainations.
""")

messages=[
    system_msg,
    HumanMessage("How do I create a Rest API?")
]
response=model.invoke(messages)
print(response.content)
```

```python
# Message Metadata

human_msg=HumanMessage(
    content="Hello!",
    name="alice",  #optional 
    id="msg_123"   #optional
)
```

```python
response=model.invoke([human_msg])
print(response.content)
```

### AI Message

```python
from langchain.messages import SystemMessage,HumanMessage,AIMessage

ai_msg=AIMessage("I'd be happy to help you with that question!")

messages=[
    SystemMessage("You are a helpful assistant"),
    HumanMessage("Can you help me?"),
    ai_msg,
    HumanMessage("Great! What's 2+2?")
]

response=model.invoke(messages)
print(response.content)
```

```python
response.usage_metadata
```

### Tool Message

```python
from langchain.messages import ToolMessage,AIMessage,HumanMessage

ai_msg=AIMessage(
    content=[],
    tool_calls=[{
        "name":"get_weather",
        "args":{"location":"bengaluru"},
        "id":"call_123"
    }]
)
weather_result="Rainy"
tool_msg=ToolMessage(
    content=weather_result,
    tool_call_id="call_123"
)
messages=[
    HumanMessage("What's weather in the Bengaluru?"),
    ai_msg,
    tool_msg
]
response=model.invoke(messages)
```

```python
print(response.content)
```
