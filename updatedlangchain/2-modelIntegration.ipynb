### GROQ Model Integration

```python
import os
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
model=init_chat_model("groq:openai/gpt-oss-120b")
model

response=model.invoke("why do parrot talk?")
response.content
```

```python
import os
from groq import Groq

client = Groq(api_key=os.environ.get("GROQ_API_KEY"))
print([m.id for m in client.models.list().data])
```

```python
from langchain_groq import ChatGroq

model = ChatGroq(
    model="openai/gpt-oss-120b"  # or "openai/gpt-oss-20b"
)

response = model.invoke("where is delhi")
response.content
```

### Streaming and Batch

```python
model.invoke("write me a 500 words paragraph on AI")
```

```python
for chunk in model.stream("write me a 500 words paragraph on AI"):
    print(f"|{chunk.content}", end="", flush=True)
```

Batch

```python
reponses=model.batch(["why do parrots have colorful feather?",
"how do airplane fly",
"what is AI?",
"where is delhi",
"who are the ipl winners",
"who won more orange caps",
"who scorred most runs in ipl"],
config={'max_concurrency':5})
for response in reponses:
    print(response)

```
